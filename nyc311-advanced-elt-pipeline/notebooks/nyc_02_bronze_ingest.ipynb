```python
from pyspark.sql.functions import current_timestamp

spark.sql("CREATE SCHEMA IF NOT EXISTS nyc311_project.bronze")

raw_path = "/Volumes/nyc311_project/default/raw_data/nyc311_raw/"

df = (spark.read
      .option("header", True)
      .csv(raw_path)
      .withColumn("_ingested_at", current_timestamp()))

df.write.format("delta").mode("overwrite") \
    .saveAsTable("nyc311_project.bronze.service_requests")

print(f"Bronze rows: {df.count()}")
```
