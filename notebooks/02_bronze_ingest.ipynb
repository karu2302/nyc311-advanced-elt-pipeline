```python
from pyspark.sql.functions import current_timestamp

raw_path = "/Volumes/olist_project/default/raw_data"
bronze_catalog = "olist_project"
bronze_schema = "bronze"

# Create the bronze schema if it doesn't exist
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {bronze_catalog}.{bronze_schema}")

file_to_table = {
    "olist_customers_dataset.csv": "customers",
    "olist_geolocation_dataset.csv": "geolocation",
    "olist_order_items_dataset.csv": "order_items",
    "olist_order_payments_dataset.csv": "order_payments",
    "olist_order_reviews_dataset.csv": "order_reviews",
    "olist_orders_dataset.csv": "orders",
    "olist_products_dataset.csv": "products",
    "olist_sellers_dataset.csv": "sellers",
    "product_category_name_translation.csv": "category_translation",
}

for filename, table_name in file_to_table.items():
    df = (spark.read
          .option("header", True)
          .option("inferSchema", True)
          .csv(f"{raw_path}/{filename}")
          .withColumn("_ingested_at", current_timestamp()))

    full_table_name = f"{bronze_catalog}.{bronze_schema}.{table_name}"
    df.write.format("delta").mode("overwrite").saveAsTable(full_table_name)
    print(f"Loaded {full_table_name}: {df.count()} rows")
```
