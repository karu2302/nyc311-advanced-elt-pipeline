```python
df = spark.read.option("header", True).csv("/Volumes/nyc311_project/default/raw_data/nyc311_raw/")
display(df)
df.printSchema()
```

```python
from pyspark.sql.functions import col, count, when

total = df.count()
for c in df.columns:
    nulls = df.filter(col(c).isNull() | (col(c) == "")).count()
    pct = round((nulls/total)*100, 1)
    if pct > 0:
        print(f"{c:<30} {nulls:>8} nulls ({pct}%)")
```
